FastAPI

FastAPI is a modern Python framework used to build APIs quickly and efficiently. It's especially useful for backend development, machine learning, and GenAI applications.
1. What is an API?
API stands for Application Programming Interface. It allows two applications to communicate with each other.

For example, imagine you build a machine learning model that predicts house prices.

FastAPI connects the user interface to your Python code or ML model.

2. Why use FastAPI?
- Fast: Designed for high-performance API development.
- Easy to learn: Uses standard Python functions and type hints.
- Automatic documentation: Generates interactive API docs.
- Data validation: Checks whether incoming data has the expected format.
- Async support: Can handle asynchronous operations.

3. Important FastAPI concepts
Concept       	Meaning
FastAPI()	      Creates your application
@app.get()	    Handles GET requests
@app.post()	    Handles POST requests
Endpoint	      A URL where your API can be accessed
Request	        Data sent to the API
Response	      Data returned by the API
Pydantic	      Validates and structures data
Uvicorn	        ASGI server used to run the application
